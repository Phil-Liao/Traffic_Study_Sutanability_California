Open 'avgts.csv' using Pandas

```python
import pandas as pd
# Open the CSV file
df = pd.read_csv("avgts.csv")

# Display the first few rows of the dataframe
print(df.head())
```

Remove the rows with no data in column "district_name"

```python
# Remove rows with missing values in 'district_name'
df.dropna(subset=['district_name'], inplace=True)

# Save the updated dataframe to the CSV file
df.to_csv("avgts.csv", index=False)

print("Updated 'avgts.csv' successfully by removing rows with no data in 'district_name'.")
print(df.head())
```

Check if there are any empty rows in "district_name"

```python
# Check for missing values in 'district_name'
missing_values = df['district_name'].isnull().sum()

if missing_values == 0:
    print("No empty rows found in 'district_name'.")
else:
    print(f"Found {missing_values} empty rows in 'district_name'.")
```

Strip so that the remaining values so that there are only number is the column "district_name"

```python
# Extract numbers from 'district_name'
df['district_name'] = df['district_name'].str.extract(r'(\d+)')

# Save the updated dataframe to the CSV file
df.to_csv("avgts.csv", index=False)

print("Updated 'district_name' column to contain only numbers and saved to 'avgts.csv'.")
print(df.head())
```

Remove the "route_name" columnn

```python
# Remove the 'route_name' column
df.drop(columns=['route_name'], inplace=True)

# Save the updated dataframe to the CSV file
df.to_csv("avgts.csv", index=False)

print("Removed 'route_name' column and saved to 'avgts.csv'.")
print(df.head())
```

Remove the "agency" columnn

```python
# Remove the 'agency' column
df.drop(columns=['agency'], inplace=True)

# Save the updated dataframe to the CSV file
df.to_csv("avgts.csv", index=False)

print("Removed 'agency' column and saved to 'avgts.csv'.")
print(df.head())
```

Remove the "OBJECTID" columnn

```python
# Remove the 'OBJECTID' column
df.drop(columns=['OBJECTID'], inplace=True)

# Save the updated dataframe to the CSV file
df.to_csv("avgts.csv", index=False)

print("Removed 'OBJECTID' column and saved to 'avgts.csv'.")
print(df.head())
```
