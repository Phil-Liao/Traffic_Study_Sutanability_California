Opening the original file hci_transportation2work_42_ct_pl_co_re_st_12-12-13-revised-ada.xlsx to create a clone t2w.xlsx for control

```python
import pandas as pd

def clone_excel_file(source_file='hci_transportation2work_42_ct_pl_co_re_st_12-12-13-revised-ada.xlsx', target_file='t2w.xlsx'):
    # Read all sheets from the source Excel file
    excel_data = pd.read_excel(source_file, sheet_name=None)
    # Write all sheets to the target Excel file
    with pd.ExcelWriter(target_file, engine='openpyxl') as writer:
        for sheet_name, df in excel_data.items():
            df.to_excel(writer, sheet_name=sheet_name, index=False)
clone_excel_file()
```

Clear all rows in t2w without a county

```python
# Load the Excel file
excel_data = pd.read_excel('t2w.xlsx', sheet_name=None)

# Remove rows where 'county_name' is missing in each sheet
for sheet_name, df in excel_data.items():
    if 'county_name' in df.columns:
        excel_data[sheet_name] = df[df['county_name'].notna()]
    else:
        excel_data[sheet_name] = df

# Save the cleaned data back to t2w.xlsx
with pd.ExcelWriter('t2w.xlsx', engine='openpyxl') as writer:
    for sheet_name, df in excel_data.items():
        df.to_excel(writer, sheet_name=sheet_name, index=False)
```

Sum up the "pop_total" and "pop_mode" for the data with the same "reportyear", "county_name", "mode"

```python
# Group by 'reportyear', 'county_name', and 'mode' and sum 'pop_total' and 'pop_mode'
# For other columns, we'll take the first value as they should be the same within each group after sorting
aggregation_functions = {
    'pop_total': 'sum',
    'pop_mode': 'sum',
    'county_fips': 'first',
    'mode_name': 'first',
    'percent': 'first'  # Keep the percent column, taking the first (all are NA)
}

mode_df_grouped = mode_df.groupby(['reportyear', 'county_name', 'mode'], as_index=False).agg(aggregation_functions)

# Reorder columns to match the original order as much as possible
# Original columns: 'reportyear', 'county_name', 'county_fips', 'mode', 'mode_name', 'pop_total', 'pop_mode', 'percent'
desired_column_order = ['reportyear', 'county_name', 'county_fips', 'mode', 'mode_name', 'pop_total', 'pop_mode', 'percent']
mode_df_grouped = mode_df_grouped[desired_column_order]

# Assign the grouped and aggregated data back to mode_df
mode_df = mode_df_grouped

# Write the updated dataframe back to the same sheet in t2w.xlsx
with pd.ExcelWriter('t2w.xlsx', engine='openpyxl', mode='a', if_sheet_exists='replace') as writer:
    mode_df.to_excel(writer, sheet_name='mode_of_transportation', index=False)

print("Data aggregated and saved to t2w.xlsx")
print(f"Number of rows after aggregation: {len(mode_df)}")
```

Recalculate the "percent" column for each row

```python
# Recalculate the 'percent' column
# Ensure 'pop_total' and 'pop_mode' are numeric and handle potential division by zero
mode_df['pop_total'] = pd.to_numeric(mode_df['pop_total'], errors='coerce')
mode_df['pop_mode'] = pd.to_numeric(mode_df['pop_mode'], errors='coerce')

# Calculate percent, handling cases where pop_total is zero or NaN to avoid division by zero
mode_df['percent'] = mode_df.apply(
    lambda row: (row['pop_mode'] / row['pop_total']) * 100
    if pd.notna(row['pop_total']) and pd.notna(row['pop_mode']) and row['pop_total'] != 0
    else pd.NA,
    axis=1
)

# Write the updated dataframe back to the same sheet in t2w.xlsx
with pd.ExcelWriter('t2w.xlsx', engine='openpyxl', mode='a', if_sheet_exists='replace') as writer:
    mode_df.to_excel(writer, sheet_name='mode_of_transportation', index=False)

print("Recalculated 'percent' column and saved to t2w.xlsx")
print(mode_df[['pop_total', 'pop_mode', 'percent']].head())
```
