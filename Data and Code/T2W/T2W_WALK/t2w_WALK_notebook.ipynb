Open "t2w_WALK.xlsx" with pandas

```python
import pandas as pd

df = pd.read_excel("t2w_WALK.xlsx")
df.head()
```

Sort the data by "district_name"

```python
df_sorted = df.sort_values(by="district_name")
df_sorted.to_excel("t2w_WALK.xlsx", index=False)
df_sorted.head()
```

Drop the column "county_name"

```python
df_sorted = df_sorted.drop(columns=["county_name"])
df_sorted.to_excel("t2w_WALK.xlsx", index=False)
df_sorted.head()
```

Clear the "percent" column

```python
df_sorted["percent"] = None
df_sorted.to_excel("t2w_WALK.xlsx", index=False)
df_sorted.head()
```

Merge the rows for each "district_name"

```python
df_merged = df_sorted.groupby("district_name", as_index=False).agg({
    **{col: "first" for col in df_sorted.columns if col not in ["pop_total", "pop_mode", "district_name"]},
    "pop_total": "sum",
    "pop_mode": "sum"
})
df_merged.to_excel("t2w_WALK.xlsx", index=False)
df_merged.head()
```

Calculate the "percent" by formula: "pop_mode"/"pop_total"*100

```python
df_merged["percent"] = (df_merged["pop_mode"] / df_merged["pop_total"]) * 100
df_merged.to_excel("t2w_WALK.xlsx", index=False)
df_merged.head()
```
