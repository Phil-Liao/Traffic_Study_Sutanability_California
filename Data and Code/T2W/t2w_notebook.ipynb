Processing of data in T2W. Note the any modification of the file without the use of this notebook requires cell [2] to be rerun for this notebook to properly.

Opening the original file hci_transportation2work_42_ct_pl_co_re_st_12-12-13-revised-ada.xlsx to create a clone t2w.xlsx for control

```python
import pandas as pd

def clone_excel_file(source_file='hci_transportation2work_42_ct_pl_co_re_st_12-12-13-revised-ada.xlsx', target_file='t2w.xlsx'):
    # Read all sheets from the source Excel file
    excel_data = pd.read_excel(source_file, sheet_name=None)
    # Write all sheets to the target Excel file
    with pd.ExcelWriter(target_file, engine='openpyxl') as writer:
        for sheet_name, df in excel_data.items():
            df.to_excel(writer, sheet_name=sheet_name, index=False)
clone_excel_file()
```

Open t2w.xlsx

```python
import pandas as pd
# Load the 'mode_of_transportation' sheet
mode_df = pd.read_excel('t2w.xlsx', sheet_name='mode_of_transportation')
```

Remove all other sheets of "t2w.xlsx" except the "mode_of_transportation" sheet

```python
from openpyxl import load_workbook

# Load the workbook
wb = load_workbook('t2w.xlsx')

# Remove all sheets except 'mode_of_transportation'
sheets_to_remove = [sheet for sheet in wb.sheetnames if sheet != 'mode_of_transportation']
for sheet in sheets_to_remove:
    std = wb[sheet]
    wb.remove(std)

# Save the workbook
wb.save('t2w.xlsx')
```

Remove the last row of data where the 'ind_id' is "END OF TABLE"

```python
# Remove the last row where 'ind_id' is "END OF TABLE" and save the file
mode_df = mode_df[mode_df['ind_id'] != "END OF TABLE"]
mode_df.to_excel('t2w.xlsx', sheet_name='mode_of_transportation', index=False)
```

Drop the columns {"ind_id", "ind_definition", "race_eth_name", "LL95CI_percent", "UL95CI_percent", "percent_se", "percent_rse", "CA_decile", "CA_RR"}

```python
# Drop the specified columns and save the updated dataframe back to the file
columns_to_drop = {"ind_id", "ind_definition", "race_eth_name", "LL95CI_percent", "UL95CI_percent", "percent_se", "percent_rse", "CA_decile", "CA_RR"}
mode_df = mode_df.drop(columns=columns_to_drop)
mode_df.to_excel('t2w.xlsx', sheet_name='mode_of_transportation', index=False)
```

Drop the columns {"region_name", "region_code"}

```python
# Drop the specified columns and save the updated dataframe back to the file
mode_df = mode_df.drop(columns=["region_name", "region_code"])
mode_df.to_excel('t2w.xlsx', sheet_name='mode_of_transportation', index=False)
```

Remove the data if the "geo_type" is not "CO"

```python
# Keep only rows where 'geotype' is 'CO' and save the updated dataframe back to the file
mode_df = mode_df[mode_df['geotype'] == 'CO']
mode_df.to_excel('t2w.xlsx', sheet_name='mode_of_transportation', index=False)
```

Drop the column "geotype"

```python
# Drop the 'geotype' column and save the updated dataframe back to the file
mode_df = mode_df.drop(columns=["geotype"])
mode_df.to_excel('t2w.xlsx', sheet_name='mode_of_transportation', index=False)
```

Remove the data if the "race_eth_code" is not 9

```python
# Keep only rows where 'race_eth_code' is 9 and save the updated dataframe back to the file
mode_df = mode_df[mode_df['race_eth_code'] == 9]
mode_df.to_excel('t2w.xlsx', sheet_name='mode_of_transportation', index=False)
```

Drop the column "race_eth_code"

```python
# Drop the 'race_eth_code' column and save the updated dataframe back to the file
mode_df = mode_df.drop(columns=["race_eth_code"])
mode_df.to_excel('t2w.xlsx', sheet_name='mode_of_transportation', index=False)
```

Drop the column "geoname"

```python
# Drop the 'geoname' column and save the updated dataframe back to the file
mode_df = mode_df.drop(columns=["geoname"])
mode_df.to_excel('t2w.xlsx', sheet_name='mode_of_transportation', index=False)
```

Drop the columns {"geotypevalue", "county_fips"}

```python
# Drop the specified columns and save the updated dataframe back to the file
mode_df = mode_df.drop(columns=["geotypevalue", "county_fips"])
mode_df.to_excel('t2w.xlsx', sheet_name='mode_of_transportation', index=False)
```

Clear data in the column "percent"

```python
# Clear data in the 'percent' column by setting all values to NaN
mode_df['percent'] = pd.NA
mode_df.to_excel('t2w.xlsx', sheet_name='mode_of_transportation', index=False)
```

Drop the "version" column

```python
# Drop the 'version' column and save the updated dataframe back to the file
mode_df = mode_df.drop(columns=["version"])
mode_df.to_excel('t2w.xlsx', sheet_name='mode_of_transportation', index=False)
```

Sort the rows of data by "county_name" then "mode" then "reportyear"

```python
# Sort the dataframe by 'county_name', then 'mode', then 'reportyear'
mode_df = mode_df.sort_values(by=["county_name", "mode", "reportyear"])
mode_df.to_excel('t2w.xlsx', sheet_name='mode_of_transportation', index=False)
```

Append a empty column "district_name" right of "county_name"

```python
# Insert an empty 'district_name' column right after 'county_name'
col_idx = mode_df.columns.get_loc("county_name") + 1
mode_df.insert(col_idx, "district_name", pd.NA)
mode_df.to_excel('t2w.xlsx', sheet_name='mode_of_transportation', index=False)
```

Remove the row if the data in "reportyear" is not "2005-2007" or "2008-2010"

```python
# Keep only rows where 'reportyear' is "2005-2007" or "2008-2010"
mode_df = mode_df[mode_df['reportyear'].isin(["2005-2007", "2008-2010"])]
mode_df.to_excel('t2w.xlsx', sheet_name='mode_of_transportation', index=False)
```

Remove the row if at least one of the values in the columns {"pop_total", "pop_mode"} is empty

```python
# Remove rows where 'pop_total' or 'pop_mode' is empty (NaN)
mode_df = mode_df.dropna(subset=["pop_total", "pop_mode"])
mode_df.to_excel('t2w.xlsx', sheet_name='mode_of_transportation', index=False)
```

Fill up the value for the "district_name" based on the "county_name" and the refernce file "caltrans_county_reference.csv"

```python
# Load the reference file
ref_df = pd.read_csv("caltrans_county_reference.csv")

# Create a mapping from county_name to district_name
county_to_district = dict(zip(ref_df["County Name"], ref_df["District"]))

# Fill the 'district_name' column based on the mapping
mode_df["district_name"] = mode_df["county_name"].map(county_to_district)

# Save the updated dataframe back to the file
mode_df.to_excel('t2w.xlsx', sheet_name='mode_of_transportation', index=False)
```

Merge the rows by adding the "pop_total" and the "pop_mode"

```python
import numpy as np

# Merge every two consecutive rows as a group, summing 'pop_total' and 'pop_mode', and updating 'reportyear'

# Ensure the dataframe is sorted as before
mode_df = mode_df.sort_values(by=["county_name", "mode", "reportyear"]).reset_index(drop=True)

# Create a list to store merged rows
merged_rows = []

# Iterate in steps of 2
for i in range(0, len(mode_df), 2):
    # If there is a pair to merge
    if i + 1 < len(mode_df):
        row1 = mode_df.iloc[i]
        row2 = mode_df.iloc[i + 1]
        # Merge the two rows
        merged_row = row1.copy()
        merged_row['pop_total'] = row1['pop_total'] + row2['pop_total']
        merged_row['pop_mode'] = row1['pop_mode'] + row2['pop_mode']
        merged_row['reportyear'] = '2005-2010'
        merged_rows.append(merged_row)
    else:
        # If odd number of rows, keep the last row as is
        merged_rows.append(mode_df.iloc[i])

# Create a new DataFrame from merged rows
merged_df = pd.DataFrame(merged_rows).reset_index(drop=True)

# Save the merged dataframe back to the file
merged_df.to_excel('t2w.xlsx', sheet_name='mode_of_transportation', index=False)

# Update mode_df for further use
mode_df = merged_df
```

Drop the column "reportyear"

```python
# Drop the 'reportyear' column and save the updated dataframe back to the file
mode_df = mode_df.drop(columns=["reportyear"])
mode_df.to_excel('t2w.xlsx', sheet_name='mode_of_transportation', index=False)
```

Calculate the "percent" by formula: "pop_mode"/"pop_total"*100

```python
# Calculate the 'percent' column as (pop_mode / pop_total) * 100
mode_df['percent'] = (mode_df['pop_mode'] / mode_df['pop_total']) * 100

# Save the updated dataframe back to the file
mode_df.to_excel('t2w.xlsx', sheet_name='mode_of_transportation', index=False)
```

Remove the data if the "mode" is not the specified r2k list

```python
r2k = ["WALK"]
#rows to keep
mode_df = mode_df[mode_df['mode'].isin(r2k)]
file_name = f't2w_{r2k[0]}.xlsx'
mode_df.to_excel(file_name, sheet_name='mode_of_transportation', index=False)
```

Move the new file to a new folder named f"T2W_{r2k}"

```python
import os
import shutil

# Create the new folder if it doesn't exist
folder_name = f"T2W_{r2k[0]}"
os.makedirs(folder_name, exist_ok=True)

# Move the file to the new folder
shutil.move(file_name, os.path.join(folder_name, file_name))
```
